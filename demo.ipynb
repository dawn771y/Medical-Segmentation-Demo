```python
import os
import urllib.request

# 1. 确保 data 文件夹存在（如果还没有，会自动创建）
os.makedirs("data", exist_ok=True)

# 2. 指定下载的目标路径，直接保存在 data 文件夹内
output_filename = "data/test_image.nii.gz"
url = "https://github.com/neurolabusc/niivue-images/raw/main/mni152.nii.gz"

print("正在下载测试医学图像到 data 目录...")
urllib.request.urlretrieve(url, output_filename)
print("下载完成！文件已保存在:", output_filename)

# 3. 验证是否能成功读取
import SimpleITK as sitk

image = sitk.ReadImage(output_filename)
print("成功读取！图像尺寸 (Size):", image.GetSize())
print("像素间距 (Spacing):", image.GetSpacing())
```
